```python
# from datasheet of cedrat piezo ppa20m: the hysteresys curve gives position vs voltage. As we are interested in the full amplitude for smaller stroke, i choose midpoint position and interpolate. 
# from illustrator: position: 2.5mum/10mm, voltage: 25V/20mm

imgtoV = 25/20  # V/mm
imgtom = 2.5e-6/10 # V/um

V0 = 0
V1 = 20
x0_up = imgtom * 43
x0_down = imgtom * 50
x1_up = imgtom * 50.7
x1_down = imgtom * 62.15


V0 = -10
V1 = 10
x0_up = imgtom * 39
x0_down = imgtom * 44
x1_up = imgtom * 46.5
x1_down = imgtom * 56



x0_mid = (x0_up + x0_down) / 2
x1_mid = (x1_up + x1_down) / 2  

slope = (x1_up - x0_up)/(V1 - V0)
print("slope at 0 V (m/V): ", slope )
print("displacement with 20 V (m): ", slope*20)

# now measured strain gauge data
dVpp  = 691e-6  # V
dxpp  = slope * 20  # m
Vdc  = 2 # V

C =  dVpp/dxpp / 2
C/2/1000

print("strin gauge calibration constant (V/m/V): ", C)
print("strain gauge calibration constant (mV/mum/V): ", C/1000)
print("strain gauge calibration constant on 50 Ohm input (mV/mum/V): ", C/1000/2)
```
