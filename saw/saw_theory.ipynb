```python
import numpy as np
from numpy.linalg import eig, inv, svd

# --------- Basic tensor/rotation helpers ---------

def voigt_to_tensor(C_voigt):
    """
    Convert 6x6 Voigt stiffness matrix (tensorial Voigt convention) to
    full 3x3x3x3 c_ijkl with standard mapping.
    """
    voigt_pairs = [(0,0), (1,1), (2,2), (1,2), (0,2), (0,1)]
    c = np.zeros((3,3,3,3), dtype=float)
    for I in range(6):
        i, j = voigt_pairs[I]
        for J in range(6):
            k, l = voigt_pairs[J]
            c[i,j,k,l] = C_voigt[I,J]
            c[j,i,k,l] = C_voigt[I,J]
            c[i,j,l,k] = C_voigt[I,J]
            c[j,i,l,k] = C_voigt[I,J]
    return c

def rotation_matrix_ST(variant="A", theta_deg=42.75):
    """
    Simple ST-like rotation: rotate about crystallographic Y by +/- theta.
    This is a standard rotated Y-cut model; you can refine later if needed.
    """
    theta = np.deg2rad(theta_deg)
    if variant.upper() == "A":
        th = theta
    else:
        th = -theta

    R_y = np.array([
        [ np.cos(th), 0.0, np.sin(th)],
        [ 0.0,        1.0, 0.0       ],
        [-np.sin(th), 0.0, np.cos(th)]
    ])
    return R_y

def rotate_c_tensor(c, R):
    """
    Rotate c_ijkl with direction cosine matrix R (3x3).
    c'_pqrs = R_{pi} R_{qj} R_{rk} R_{sl} c_ijkl
    """
    c_rot = np.zeros_like(c)
    for p in range(3):
        for q in range(3):
            for r in range(3):
                for s in range(3):
                    val = 0.0
                    for i in range(3):
                        for j in range(3):
                            for k in range(3):
                                for l in range(3):
                                    val += (R[p,i] * R[q,j] *
                                            R[r,k] * R[s,l] * c[i,j,k,l])
                    c_rot[p,q,r,s] = val
    return c_rot

# --------- Stroh construction (3D sagittal) ---------

def build_Q_R_T(c, rho, v):
    """
    Build Q, R, T (3x3) for 3D anisotropic Rayleigh problem in the
    (x1 = propagation, x2 = depth, x3 = out-of-plane) frame.
    """
    Q = np.zeros((3,3), dtype=complex)
    R = np.zeros((3,3), dtype=complex)
    T = np.zeros((3,3), dtype=complex)

    # indices: 0->x1, 1->x2, 2->x3
    for i in range(3):
        for k in range(3):
            Q[i,k] = c[i,1,k,1]          # c_i2k2
            R[i,k] = c[i,1,k,0]          # c_i2k1
            T[i,k] = c[i,0,k,0]          # c_i1k1
    T -= rho * v**2 * np.eye(3)
    return Q, R, T

def stroh_matrix(Q, R, T):
    """
    6x6 Stroh matrix:
      N = [[-T^-1 R^T,      -T^-1],
           [Q - R T^-1 R^T, -R T^-1]]
    """
    Ti = inv(T)
    upper_left  = -Ti @ R.T
    upper_right = -Ti
    lower_left  = Q - R @ Ti @ R.T
    lower_right = -R @ Ti
    N = np.block([[upper_left,  upper_right],
                  [lower_left,  lower_right]])
    return N

# --------- Build Rayleigh mode and H ---------

def build_rayleigh_mode(c_rot, rho, vR, k=1.0, x2_max=6.0, nx=800):
    """
    Construct the Rayleigh mode (u1,u2,u3)(x2) for given c_rot, rho, vR.
    Normalizes so that |u(0)| ~ 1.
    """
    Q, R, T = build_Q_R_T(c_rot, rho, vR)
    N = stroh_matrix(Q, R, T)

    vals, vecs = eig(N)
    p = vals

    # pick decaying modes: Im(p) > 0
    idx = np.where(np.imag(p) > 0)[0]
    if len(idx) != 3:
        raise RuntimeError(f"Expected 3 decaying modes, got {len(idx)}")

    a_modes = vecs[0:3, idx]  # displacement parts (3x3)
    b_modes = vecs[3:6, idx]  # traction parts (3x3)

    # traction-free boundary: B c ≈ 0 -> smallest singular vector of B
    U, S, Vh = svd(b_modes)
    c_coeff = Vh[-1, :]
    # surface displacement (3 components)
    u0 = a_modes @ c_coeff
    U0_mag = np.sqrt(np.real(np.vdot(u0, u0)))
    c_coeff /= U0_mag
    u0 = a_modes @ c_coeff  # normalized: |u0| ≈ 1

    # depth grid and mode profile
    x2 = np.linspace(0.0, x2_max, nx)
    u = np.zeros((3, nx), dtype=complex)
    for m in range(3):
        pm = p[idx[m]]
        a_m = a_modes[:, m]
        c_m = c_coeff[m]
        u += np.outer(a_m * c_m, np.exp(1j * pm * k * x2))

    return x2, u, u0

def compute_strains_2d(u, x2, k=1.0):
    """
    2D sagittal approximation: we keep in-plane derivatives for u1,u2.
    u = [u1,u2,u3]; u3 is small but kept in kinetic energy.
    """
    u1, u2, u3 = u[0,:], u[1,:], u[2,:]
    nx = len(x2)
    dx = x2[1] - x2[0]

    # derivatives wrt x1: du/dx1 = i k u
    u1_x1 = 1j * k * u1
    u2_x1 = 1j * k * u2

    # derivatives wrt x2: finite differences
    u1_x2 = np.zeros_like(u1, dtype=complex)
    u2_x2 = np.zeros_like(u2, dtype=complex)
    u1_x2[1:-1] = (u1[2:] - u1[:-2]) / (2*dx)
    u2_x2[1:-1] = (u2[2:] - u2[:-2]) / (2*dx)
    u1_x2[0]  = (u1[1]   - u1[0])   / dx
    u1_x2[-1] = (u1[-1]  - u1[-2])  / dx
    u2_x2[0]  = (u2[1]   - u2[0])   / dx
    u2_x2[-1] = (u2[-1]  - u2[-2])  / dx

    # non-zero strain components in sagittal plane
    eps11 = u1_x1
    eps22 = u2_x2
    eps12 = 0.5 * (u1_x2 + u2_x1)
    return eps11, eps22, eps12

def compute_H_total(c_rot, rho, vR, x2, u, k=1.0):
    """
    Compute H for the total surface amplitude:
      H = ∫ W(x2) dx2  (for k=1, |u(0)|=1).
    """
    eps11, eps22, eps12 = compute_strains_2d(u, x2, k=k)

    nx = len(x2)
    # Voigt strain vector [11,22,33,23,13,12] (only 11,22,12 non-zero here)
    eps_voigt = np.zeros((6, nx), dtype=complex)
    eps_voigt[0,:] = eps11
    eps_voigt[1,:] = eps22
    eps_voigt[5,:] = eps12

    # recover Voigt stiffness C_rot from c_rot
    voigt_pairs = [(0,0),(1,1),(2,2),(1,2),(0,2),(0,1)]
    C_rot_voigt = np.zeros((6,6), dtype=float)
    for I in range(6):
        i,j = voigt_pairs[I]
        for J in range(6):
            k1,l = voigt_pairs[J]
            C_rot_voigt[I,J] = c_rot[i,j,k1,l]

    sigma_voigt = C_rot_voigt @ eps_voigt

    # kinetic energy density: 1/4 rho omega^2 |u|^2
    omega = vR * k
    u1, u2, u3 = u[0,:], u[1,:], u[2,:]
    u_mag2 = np.abs(u1)**2 + np.abs(u2)**2 + np.abs(u3)**2
    W_kin = 0.25 * rho * omega**2 * u_mag2

    # strain energy density: 1/4 sigma_ij eps_ij*
    W_str = 0.25 * (
        sigma_voigt[0,:] * np.conj(eps_voigt[0,:]) +
        sigma_voigt[1,:] * np.conj(eps_voigt[1,:]) +
        sigma_voigt[5,:] * np.conj(eps_voigt[5,:])
    )

    W = W_kin + np.real(W_str)
    E = np.trapz(W, x2)   # k=1, |u(0)| ≈ 1 → H ≈ E
    return E

def compute_H_components(H_total, u0):
    """
    Given H_total (for U = |u(0)|) and the surface displacement vector u0,
    compute H_parallel and H_perp using the exact complex ratio r = u2/u1.
    """
    u1_0, u2_0, u3_0 = u0
    r = u2_0 / u1_0
    H_par  = H_total * (1.0 + np.abs(r)**2)
    H_perp = H_total * (1.0 + 1.0 / (np.abs(r)**2))
    return H_par, H_perp, r

# --------- Main: run for both ST variants, Royer constants ---------

def main():
    # Royer α-quartz elastic constants (units: 10^10 N/m^2 in the book)
    # Convert to Pa: multiply by 1e10.
    C11 = 8.67e10
    C12 = 0.70e10
    C13 = 1.19e10
    C33 = 10.72e10
    C44 = 5.79e10
    C14 = -1.79e10
    C66 = (C11 - C12) / 2.0

    C_base = np.array([
        [C11, C12, C13,   0,   0, C14],
        [C12, C11, C13,   0,   0,-C14],
        [C13, C13, C33,   0,   0,   0],
        [  0,   0,   0, C44, C14,   0],
        [  0,   0,   0, C14, C44,   0],
        [C14,-C14,  0,   0,   0, C66]
    ], dtype=float)

    rho = 2650.0          # kg/m^3
    vR  = 3158.0          # m/s, ST-X Rayleigh R3
    k   = 1.0             # choose k=1 → H = E for normalized mode

    for variant in ["A", "B"]:
        print(f"\n=== ST-cut variant {variant} (elastic only, Royer constants) ===")
        R = rotation_matrix_ST(variant=variant, theta_deg=42.75)
        c_tensor = voigt_to_tensor(C_base)
        c_rot = rotate_c_tensor(c_tensor, R)

        x2, u, u0 = build_rayleigh_mode(c_rot, rho, vR, k=k, x2_max=6.0, nx=800)
        H_total   = compute_H_total(c_rot, rho, vR, x2, u, k=k)
        H_par, H_perp, r = compute_H_components(H_total, u0)

        U0 = np.sqrt(np.real(np.vdot(u0, u0)))

        print(f"  |u(0)| (normalization) ≈ {U0:.3f}")
        print(f"  u1(0) ≈ {u0[0]:.3e}, u2(0) ≈ {u0[1]:.3e}, u3(0) ≈ {u0[2]:.3e}")
        print(f"  r = u2/u1 ≈ {r:.3f}")
        print(f"  H_total   ≈ {H_total:.3e} N/m^2")
        print(f"  H_parallel (along propagation) ≈ {H_par:.3e} N/m^2")
        print(f"  H_perp (normal to surface)     ≈ {H_perp:.3e} N/m^2")


if __name__ == "__main__":
    main()
```

```python
def lorentzian1(p, x):

    c0, x0, fullwidth, rate = p
    
    return rate*(1 -  c0*(fullwidth/2)**2 / ((x - x0)**2 + (fullwidth/2)**2))
```
